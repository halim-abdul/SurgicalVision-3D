# Phase recognition demo
Synthetic clip inference to verify tensor flow without patient data.

```python
import torch
from surgicalvision3d.phase.model import PhaseRecognitionModel
model = PhaseRecognitionModel(num_phases=7).eval()
clip = torch.randn(1, 8, 3, 128, 128)
with torch.no_grad(): logits = model(clip)
prob = logits.softmax(-1)
print(prob.shape, prob.max(-1).values.mean().item())
```
