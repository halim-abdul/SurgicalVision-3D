# Geometry demo
Create a synthetic depth surface and back-project it to 3D.

```python
import numpy as np
from surgicalvision3d.geometry.pointcloud import depth_to_points
h,w=80,120; yy,xx=np.mgrid[:h,:w]; depth=0.35+0.02*np.sin(xx/12)*np.cos(yy/10)
pts=depth_to_points(depth,220,220,w/2,h/2)
print(pts.shape, pts.min(0), pts.max(0))
```
