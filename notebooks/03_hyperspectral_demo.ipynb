# Hyperspectral tissue demo
Synthetic spectra are used to demonstrate calibration, corruption and model interfaces.

```python
import numpy as np, torch
from surgicalvision3d.hyperspectral.preprocess import reflectance_calibration
from surgicalvision3d.hyperspectral.model import SpectralSpatialNet
raw=np.random.default_rng(7).uniform(2,8,(16,16,31)); dark=np.ones_like(raw); white=np.ones_like(raw)*10
cube=reflectance_calibration(raw,white,dark)
x=torch.tensor(cube.transpose(2,0,1),dtype=torch.float32)[None,None]
print(cube.shape, SpectralSpatialNet(4)(x).shape)
```
