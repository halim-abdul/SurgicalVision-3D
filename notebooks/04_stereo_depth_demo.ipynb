# Stereo depth demo
Synthetic shifted textures demonstrate disparity-to-depth behavior without medical data.

```python
from surgicalvision3d.stereo.synthetic import shifted_pair
from surgicalvision3d.stereo.depth import compute_sgbm, disparity_to_depth
left,right=shifted_pair(160,240,8)
disp=compute_sgbm(left,right,num_disparities=64,block_size=5)
depth=disparity_to_depth(disp,500.0,0.05)
print('median disparity', float(__import__('numpy').nanmedian(disp[disp>0])))
```
