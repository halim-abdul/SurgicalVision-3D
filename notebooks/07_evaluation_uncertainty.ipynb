# Evaluation and uncertainty
Synthetic probabilities demonstrate calibration metrics and uncertainty summaries.

```python
import numpy as np, torch
from surgicalvision3d.evaluation.calibration import expected_calibration_error
from surgicalvision3d.evaluation.uncertainty import predictive_entropy
p=np.array([[.8,.1,.1],[.4,.35,.25],[.05,.05,.9]])
y=np.array([0,1,2]); conf=p.max(1); pred=p.argmax(1)
print('ECE', expected_calibration_error(conf,pred==y))
print('entropy', predictive_entropy(torch.tensor(p)).numpy())
```
