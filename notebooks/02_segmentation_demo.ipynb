# Segmentation demo
Tensor-shape, loss and metric smoke test using synthetic masks.

```python
import torch
from surgicalvision3d.segmentation.unet import UNet
from surgicalvision3d.segmentation.losses import DiceCrossEntropyLoss
x=torch.randn(2,3,128,128); y=torch.randint(0,3,(2,128,128))
model=UNet(3); logits=model(x); loss=DiceCrossEntropyLoss()(logits,y)
print(logits.shape, float(loss))
```
